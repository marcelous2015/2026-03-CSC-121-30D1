```python
# NAME: roster.py

# IMPORT STATEMENTS
# No imports are needed for this program.

# FUNCTION DEFINITIONS

# Define a function that accepts the roster list.
def add_players(roster):
    # Ask the user to enter a player name.
    player = input("Enter player name (or 'quit' to stop): ")

    # Continue collecting names until the user types quit.
    while player.lower() != "quit":
        # Add the player's name to the roster list.
        roster.append(player)

        # Ask for the next player name.
        player = input("Enter player name (or 'quit' to stop): ")


# Define a fruitful function to search for a player.
def is_on_roster(roster, name):
    # Check each player in the roster.
    for player in roster:
        # Return True if the player matches the search name.
        if player == name:
            return True

    # Return False if the player is not found.
    return False


# PROGRAM LOGIC

# Create an empty roster list.
roster = []

# Call the function to collect player names.
add_players(roster)

# Ask the user which player they want to search for.
search_name = input("Who would you like to search for? ")

# Store the Boolean result returned by the search function.
result = is_on_roster(roster, search_name)

# Display whether the player was found.
if result:
    print("Result: Player found!")
else:
    print("Result: Player not found.")
```
