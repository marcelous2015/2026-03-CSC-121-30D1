```python
# NAME: gradebook.py

# IMPORT STATEMENTS


# FUNCTION DEFINITIONS

# Define add_students to collect student names and grades.
def add_students(gradebook):
    # Continue asking for student names until the user enters "done".
    while True:
        name = input("Enter student name (or 'done'): ")

        # Stop collecting students when the user enters "done".
        if name.lower() == "done":
            break

        # Continue asking until a valid grade between 0 and 100 is entered.
        while True:
            try:
                grade = float(input(f"Enter {name}'s grade: "))

                # Accept only grades between 0 and 100.
                if 0 <= grade <= 100:
                    break
                else:
                    print("Please enter a grade between 0 and 100.")
            except ValueError:
                print("Please enter a valid number.")

        # Store the student's name and grade in the dictionary.
        gradebook[name] = grade


# Define calculate_average to calculate and return the class average.
def calculate_average(gradebook):
    # Check whether the dictionary contains any students.
    if len(gradebook) == 0:
        return 0.0

    # Calculate the average using all dictionary values.
    average = sum(gradebook.values()) / len(gradebook)

    # Return the calculated average.
    return average


# PROGRAM LOGIC

# Create an empty dictionary called gradebook.
gradebook = {}

# Call add_students to collect student names and grades.
add_students(gradebook)

# Display the final class report.
print("\n--- Class Report ---")

# Display each student's name and grade.
for name, grade in gradebook.items():
    print(f"{name}: {grade:g}")

# Calculate and display the class average to two decimal places.
class_average = calculate_average(gradebook)
print(f"Class Average: {class_average:.2f}")
```
