```python
# NAME: rectangle.py

# IMPORT STATEMENTS


# FUNCTION DEFINITIONS
# Accept length and width, calculate area and perimeter,
# and return both results together in a tuple.
def calculate_rectangle(length, width):
    area = length * width
    perimeter = 2 * (length + width)
    return (area, perimeter)


# PROGRAM LOGIC
print(calculate_rectangle(5, 10))
```
