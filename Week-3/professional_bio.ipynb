```python
# Professional Bio Generator
# Week 3 Assignment

# STRUCTURE & PSEUDOCODE
# 1. Assign values to name, goal, birth_year, and current_year.
# 2. Calculate age by subtracting birth_year from current_year.
# 3. Create a border that is exactly 25 characters long using string multiplication.
# 4. Print the top border.
# 5. Print the bio sentences.
# 6. Print the bottom border.
```

```python
name = "Jordan"
goal = "become a software developer"
birth_year = 2004
current_year = 2026

age = current_year - birth_year

border = "=" * 25

print(border)
print("My name is " + name + " and I am " + str(age) + " years old.")
```
