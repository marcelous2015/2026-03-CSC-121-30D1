```python
# NAME: journal.py

# IMPORT STATEMENTS


# FUNCTION DEFINITIONS

# Saves a new journal entry to the end of the file
def write_entry(filename, entry):
    # Open the journal in append mode
    journal_file = open(filename, 'a')

    # Write the entry followed by a newline
    journal_file.write(entry + "\n")

    # Close the file
    journal_file.close()


# Reads and displays all saved journal entries
def read_journal(filename):
    try:
        # Open the journal in read mode
        journal_file = open(filename, 'r')

        # Read the entire journal
        contents = journal_file.read()

        # Close the file
        journal_file.close()

        # Display the journal entries
        print("--- Your Journal ---")
        print(contents)

    except FileNotFoundError:
        print("Your journal is empty. No entries found.")


# PROGRAM LOGIC

# Set the journal filename
filename = "journal.txt"

# Keep the journal app running until the user quits
while True:
    choice = input("(W)rite, (R)ead, or (Q)uit? ").lower()

    # Check if the user wants to write an entry
    if choice == "w":
        entry = input("Enter your entry: ")
        write_entry(filename, entry)
        print("Entry saved.")

    # Check if the user wants to read the journal
    elif choice == "r":
        read_journal(filename)

    # Exit the program
    elif choice == "q":
        print("Goodbye!")
        break

    # Handle an invalid choice
    else:
        print("Invalid choice. Please try again.")
```
