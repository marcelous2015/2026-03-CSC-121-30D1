```python
# IMPORT STATEMENTS
# (None)

# FUNCTION DEFINITIONS
def print_bio(name, age, goal):
    print("*" * 25)
    print(f"Name: {name}, Age: {age}")
    print(f"Goal: {goal}")
    print("*" * 25)

# PROGRAM LOGIC
print_bio("Jordan", 22, "Learn Python.")
```
