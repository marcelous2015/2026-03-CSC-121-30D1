```python
# Define the calculate_ticket function
def calculate_ticket(age, hour):
    # Determine the base price using the customer's age
    if age <= 12 or age >= 65:
        price = 8.00
    else:
        price = 12.00

    # Apply a $2 matinee discount if the movie starts before 17
    if hour < 17:
        price -= 2.00

    # Print the final ticket price
    print(f"Your ticket price is ${price:.2f}")


# Get the customer's age
age = int(input("Enter age: "))

# Get the movie start hour in 24-hour format
hour = int(input("Enter movie hour (0-23): "))

# Call calculate_ticket using the user's inputs
calculate_ticket(age, hour)
```
