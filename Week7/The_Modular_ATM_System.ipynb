```python

# IMPORT STATEMENTS

# No imports are needed for this program.


# FUNCTION DEFINITIONS

# Define verify_pin() with no parameters.
# Use a while loop to ask for the PIN until it equals "1234".
# Return True when the correct PIN is entered.

# Define the function that verifies the user's PIN.
def verify_pin():
    # Keep asking until the correct PIN is entered.
    while True:
        # Ask the user to enter their PIN.
        pin = input("Enter PIN: ")

        # Check whether the PIN is correct.
        if pin == "1234":
            # Confirm that the PIN was accepted.
            print("PIN Verified.")

            # Return True after successful verification.
            return True

        # Tell the user the PIN was incorrect.
        print("Incorrect PIN. Try again.")


# Define withdraw(balance, amount).
# Check whether the withdrawal amount is less than or equal to the balance.
# Return the new balance if there is enough money.
# Otherwise, print "Insufficient funds" and return the original balance.

# Define the function that processes a withdrawal.
def withdraw(balance, amount):
    # Check whether the requested amount is within the balance.
    if amount <= balance:
        # Confirm the withdrawal and return the new balance.
        print("Withdrawal successful.")
        return balance - amount

    # Inform the user that there is not enough money.
    print("Insufficient funds")

    # Return the original balance unchanged.
    return balance


# PROGRAM LOGIC

# Set the starting account balance to 500.
current_balance = 500

# Verify the PIN before allowing a withdrawal.
if verify_pin():
    # Ask for the withdrawal amount and convert it to a float.
    amount = float(input("Enter withdrawal amount: "))

    # Update the balance using the returned value.
    current_balance = withdraw(current_balance, amount)

    # Display the final account balance.
    print(f"Current Balance: ${current_balance}")


```
