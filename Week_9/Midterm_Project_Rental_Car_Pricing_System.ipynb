```python
# NAME: rentalcar.py

# IMPORT STATEMENTS


# FUNCTION DEFINITIONS

# Validate the classification code.
def validate_classification(code):
    return code in ("B", "S", "L")


# Calculate the number of miles that will be charged.
def calculate_charged_miles(total_miles):
    return max(0, total_miles - 100)


# Calculate the mileage charge.
def calculate_mileage_charge(total_miles):
    charged_miles = calculate_charged_miles(total_miles)
    return charged_miles * 0.35


# Calculate the total rental bill and return the result.
# This function does not print anything.
def calculate_bill(classification, days, total_miles):
    daily_rates = {
        "B": 40,
        "S": 60,
        "L": 80
    }

    rental_cost = daily_rates[classification] * days
    mileage_cost = calculate_mileage_charge(total_miles)

    return rental_cost + mileage_cost


# PROGRAM LOGIC

# Ask for a valid classification code.
classification = input("Enter classification code (B, S, L): ").strip().upper()

while not validate_classification(classification):
    print("Invalid classification. Please enter B, S, or L.")
    classification = input("Enter classification code (B, S, L): ").strip().upper()


# Collect the odometer readings and rental days.
start_odometer = int(input("Start Odometer: "))
end_odometer = int(input("End Odometer: "))
days = int(input("Days Rented: "))


# Calculate the total miles driven.
total_miles = end_odometer - start_odometer

# Calculate the miles that will be charged.
charged_miles = calculate_charged_miles(total_miles)

# Call the fruitful function to get the total bill.
total_bill = calculate_bill(classification, days, total_miles)


# Match the classification code to its full name.
classification_names = {
    "B": "Budget",
    "S": "Standard",
    "L": "Luxury"
}


# Display the final rental summary.
print("\n--- RENTAL SUMMARY ---")
print("Classification:", classification_names[classification])
print("Total Miles:", total_miles)
print("Charged Miles:", charged_miles)
print(f"Total Bill: ${total_bill:.2f}")
```
