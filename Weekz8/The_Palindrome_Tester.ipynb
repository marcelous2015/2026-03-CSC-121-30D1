```python
# IMPORT STATEMENTS

# No imports are needed because Python supports string slicing.


# FUNCTION DEFINITIONS

# Define a fruitful function that accepts a word and ignores capitalization.
def is_palindrome(word):
    # Convert the word to lowercase.
    word = word.lower()

    # Reverse the word using string slicing.
    reversed_word = word[::-1]

    # Return True if the word matches its reverse; otherwise return False.
    return word == reversed_word


# PROGRAM LOGIC

# Ask the user to enter a word.
word = input("Enter a word: ")

# Call the function and store its returned Boolean value.
result = is_palindrome(word)

# Print a message based on the returned value.
if result:
    print(word, "is a palindrome.")
else:
    print(word, "is not a palindrome.")
```
